```python
import numpy as np
from matplotlib import pyplot as plt 
```

```python
x=np.linspace(0,5,11)
y1=x**2
y2=x**3

frutas=['manzanas','platanos','peras']
porcentajes =[30,50,20]


categorias =['A','B','C','D']
valores=[30,50,20]


plt.subplot(2,2,1)
plt.plot(x,y1,'bo-')
plt.title('funcion cuadratica')
plt.subplot(2,2,2)#ultimo numero casilla
plt.plot(x,y2,'go-')
plt.title('funcion cubica')
plt.subplot(2,2,3)
plt.pie(porcentajes,labels=frutas,autopct='-51,1f%%')
plt.title('grafico de pastel')
plt.tight_layout(h_pad=0.5,w_pad=5.5) #separar graficos 

plt.show()
```
