```python
df["species"].value_counts()
```

```python
import matplotlib.pyplot as plt
subconjunto=df.loc[0:29,:]
plt.figure(figsize=(8,4))
plt.plot(subconjunto.index, subconjunto["sepal_length_cm"])
plt.title('longitud del sepalo')
plt.xlabel('indice')
plt.ylabel('longitud')
```

```python
import matplotlib.pyplot as plt
subconjunto=df.loc[0:29,:]
plt.figure(figsize=(8,4))
plt.plot(subconjunto.index, subconjunto["sepal_length_cm"], marker='o',linestyle='--',color='purple')
plt.title('longitud del sepalo')
plt.xlabel('indice')
plt.ylabel('longitud')
plt.grid(alpha=0.3)
plt.show
```

```python
promedio_sepal_length=df.groupby('species')["sepal_length_cm"].mean()
plt.bar(promedio_sepal_length.index,promedio_sepal_length.values)
plt.xlabel("especie")
plt.ylabel("promedio")
plt.show()
```

```python
promedio_sepal_length=df.groupby('species')["sepal_length_cm"].mean().sort_values()
plt.barh(promedio_sepal_length.index,promedio_sepal_length.values)
plt.xlabel("especie")
plt.ylabel("promedio")
plt.show()
```

```python
counts = df["species"].value_counts()
counts
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index, autopct ='%1.1f%%')
plt.title("proporciones")
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["sepal_length_cm"],df["sepal_width_cm"])
plt.title("long vs ancho")
plt.xlabel("long")
plt.ylabel("ancho")
plt.show
```
