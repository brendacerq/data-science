```python
class Cuenta:
    def __init__(self,saldo:float):
        self.saldo = saldo
    def depositar(self,monto:float):
        self.saldo += monto 
        
```

```python
cuenta=Cuenta(1000)
print(cuenta.saldo)
```

```python
cuenta.depositar(1000)
print(cuenta.saldo)
```

```python
class Persona:
    def __init__(self, nombre:str, edad:int)-> None:
        self.nombre = nombre
        self.edad= edad 

    def saludar (self):
        print ("Hola, soy {self.nombre} y tengo {self.edad} años")
    
# crear objetos 
jorge =Persona("jorge",10)
maria=Persona("Maria",20)
Luis=Persona("Luis",15)

jorge.saludar()
maria.saludar()
Luis.saludar()
```

```python
class Rectangulo:
    def __init__(self, base: float, altura: float) -> None:
        self.base = base
        self.altura = altura

    def area(self) -> float:
        return self.base * self.altura

    def perimetro(self) -> float:
        return 2 * (self.base + self.altura)

    def __str__(self) -> str:
        return f"Rectángulo de base: {self.base} y altura: {self.altura}"


# Ejemplo de uso
mi_rectangulo = Rectangulo(3, 5)

print(mi_rectangulo.area())       # Salida: 15
print(mi_rectangulo)              # Salida: Rectángulo de base: 3 y altura: 5
print(mi_rectangulo.perimetro())  # Salida: 16
```
