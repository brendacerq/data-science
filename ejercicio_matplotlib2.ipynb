```python
import matplotlib.pyplot as plt
subconjunto=df.loc[0:29,:]
plt.figure(figsize=(8,4))
plt.plot(subconjunto.index, subconjunto["sepal_length_cm"], marker='o',linestyle='--',color='purple')
plt.title('longitud del sepalo')
plt.xlabel('indice')
plt.ylabel('longitud')
plt.grid(alpha=0.3)
plt.show
```

```python
promedio_sepal_length=df.groupby('species')["sepal_length_cm"].mean()
plt.bar(promedio_sepal_length.index,promedio_sepal_length.values)
plt.xlabel("especie")
plt.ylabel("promedio")
plt.show()
```

```python
promedio_sepal_length=df.groupby('species')["sepal_length_cm"].mean().sort_values()
plt.barh(promedio_sepal_length.index,promedio_sepal_length.values)
plt.xlabel("especie")
plt.ylabel("promedio")
plt.show()
```

```python
counts = df["species"].value_counts()
counts
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index, autopct ='%1.1f%%')
plt.title("proporciones")
```

```python
## 09
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("iris_dataset.csv")

plt.figure(figsize=(7, 4))
plt.hist(
    df["sepal_length_cm"],
    bins=10,
    edgecolor="black",
    color="skyblue",
    linewidth=1.2,
)
plt.title("Distribución de la longitud del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")
plt.show()
```

```python
## 10
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 5 bins
axes[0].hist(
    df["petal_length_cm"],
    bins=5,
    edgecolor="black",
    color="lightcoral",
    linewidth=1.2,
)
axes[0].set_title("Petal Length (5 bins)")
axes[0].set_xlabel("Longitud del pétalo (cm)")
axes[0].set_ylabel("Frecuencia")

# 20 bins
axes[1].hist(
    df["petal_length_cm"],
    bins=20,
    edgecolor="black",
    color="lightgreen",
    linewidth=1.2,
)
axes[1].set_title("Petal Length (20 bins)")
axes[1].set_xlabel("Longitud del pétalo (cm)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
#11 
species_names = df["species"].unique()
data_by_species = [
    df[df["species"] == sp]["sepal_length_cm"] for sp in species_names
]

plt.figure(figsize=(7, 5))
plt.boxplot(data_by_species, tick_labels=species_names)
plt.title("Longitud del sépalo por especie")
plt.xlabel("Especie")
plt.ylabel("Longitud del sépalo (cm)")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
##12
numeric_cols = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm",
]
short_labels = ["Sepal L.", "Sepal W.", "Petal L.", "Petal W."]

plt.figure(figsize=(8, 5))
plt.boxplot([df[col] for col in numeric_cols], tick_labels=short_labels)
plt.title("Comparación de variables numéricas")
plt.ylabel("Centímetros (cm)")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
##13
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Panel izquierdo: Scatter plot
axes[0].scatter(
    df["sepal_length_cm"],
    df["sepal_width_cm"],
    alpha=0.8,
    c="steelblue",
    edgecolor="k",
)
axes[0].set_title("Dispersión: Longitud vs. Ancho de Sépalo")
axes[0].set_xlabel("Longitud del sépalo (cm)")
axes[0].set_ylabel("Ancho del sépalo (cm)")
axes[0].grid(True, linestyle="--", alpha=0.5)

# Panel derecho: Histograma
axes[1].hist(
    df["sepal_length_cm"],
    bins=10,
    edgecolor="black",
    color="orange",
    alpha=0.85,
)
axes[1].set_title("Distribución: Longitud de Sépalo")
axes[1].set_xlabel("Longitud del sépalo (cm)")
axes[1].set_ylabel("Frecuencia")
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
