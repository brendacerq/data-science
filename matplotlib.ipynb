```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]

plt.bar(categorias, valores, color='#B628FA')
plt.show()
```

```python
plt.barh(categorias,valores,color='green')
plt.xlabel('Frecuencia')
plt.ylabel('categorias')
plt.title('Grafico de barras')
plt.show()
```

```python
#grafico pastel o pie chart 
frutas=['manzana','platano','peras']
porcentajes=[30,50,30]

plt.pie(porcentajes, labels=frutas, autopct = '%1.1f%%')
plt.title('grafico de pastel')
plt.show()
```

```python
# grafico de dispersion (scatter plot)

#sinergia de variables

import numpy as np 
x=np.random.rand(50)
y=np.random.rand(50)

plt.scatter(x,y,color='blue')
plt.xlabel('variable(x)')
plt.ylabel('variable(y)')
plt.title('grafico de dispersion')
plt.show ()
```

```python
import matplotlib.pyplot as plt
import numpy as np

datos= np.random.randn(1000)
plt.hist(datos,bins=20, color ='orange',edgecolor = 'black')
plt.xlabel=('valor')
plt.ylabel('frecuencia')
plt.title('histograma')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import importlib

# Genera 3 grupos de 100 datos con media 0 y desviaciones estándar 1, 2 y 3
data_box = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.boxplot(data_box, orientation='vertical', patch_artist=True)
plt.xlabel('grupo')
plt.ylabel('valor')
plt.title('grafico boxplot')
plt.show()
```
