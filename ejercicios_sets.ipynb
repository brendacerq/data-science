```python
#1
estudiantes_python = {
"Ana", "Luis", "María", "Carlos", "Sofía", "Diego"
}
estudiantes_javascript = {
"María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria"
}
#operacion de interseccion 
print ("Estudiantes en ambos cursos", estudiantes_python & estudiantes_javascript)
#operacion diferencia
print("estudiantes solo en phyton", estudiantes_python - estudiantes_javascript)
print("estudiantes solo en java",  estudiantes_javascript - estudiantes_python)
print("todos los estudiantes",  estudiantes_javascript | estudiantes_python)

```

```python

almacen_arequipa = {
"laptop", "teclado", "mouse", "monitor", "impresora", "audifonos"
}
almacen_lima = {
"laptop", "mouse", "monitor", "tablet", "webcam", "parlantes"
}
print("Los productos disponibles en ambos almacenes",  almacen_arequipa & almacen_lima)
print("Los productos exclusivos del almacén de Arequipa",  almacen_arequipa - almacen_lima)
print ("Los productos exclusivos del almacén de Lima", almacen_lima-almacen_arequipa)
print ("Todos los productos disponibles sin repetir", almacen_lima | almacen_arequipa)
almacen_arequipa.issubset(almacen_lima)


```

```python
permisos_usuario = {
"leer", "editar", "comentar", "descargar"
}
permisos_requeridos = {
"leer", "editar", "eliminar", "publicar"
}
# Si el usuario posee todos los permisos requeridos.
print (permisos_usuario.isdisjoint(permisos_requeridos))
# Los permisos que le faltan.
print("Permisos que faltan:", permisos_requeridos-permisos_usuario)
# Los permisos adicionales que tiene.
print("Permisos adicionales:", permisos_usuario-permisos_requeridos)
# La cantidad total de permisos diferentes.
print("Permisos diferentes:", permisos_usuario ^ permisos_requeridos)
```

```python
numeros_a = {2, 4, 6, 8, 10, 12, 14}
numeros_b = {4, 6, 9, 10, 13, 14, 16}
numeros_c = {1, 4, 6, 10, 11, 14, 18}

# Los números presentes en los tres conjuntos.
print ("Los números presentes en los tres conjuntos:", numeros_a&numeros_b&numeros_c)
# Los números presentes en numeros_a y numeros_b, pero no en numeros_c.
print ("Los números presentes en numeros_a y numeros_b, pero no en numeros_c:", numeros_a & numeros_b - numeros_c)

# Los números que solo pertenecen a numeros_a.
print ("Los números que solo pertenecen a numeros_a:", numeros_a-(numeros_b|numeros_c))

# Todos los números diferentes.
print("todos los numeros diferentes:", (numeros_a^numeros_b) | (numeros_b^numeros_c) | numeros_c^numeros_a)
```
